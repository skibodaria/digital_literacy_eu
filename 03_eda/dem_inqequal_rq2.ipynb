Breaking down that data by key demographics like age, gender, education, and geography (settlement type)

Example of the code i will run here:


```python
# Create a combined variable column in Pandas
df['variable_name'] = df['indicator_code'] + '__' + df['ind_type']

# Pivot the data
# This creates one row per country per year, and a separate column for EVERY demographic mix
matrix_df = df.pivot(
    index=['country_code', 'year'], 
    columns='variable_name', 
    values='indicator_value'
)

print(matrix_df.columns)
# Your columns will look beautifully organized like this:
# 'I_DSK2_X__Y16_24_M'  -> Above-basic skills for Young Males
# 'I_DSK2_X__Y16_24_F'  -> Above-basic skills for Young Females
# 'I_DSK2_X__ED5_8_M'   -> Above-basic skills for Highly Educated Males
# 'I_DSK2_X__ED5_8_F'   -> Above-basic skills for Highly Educated Females
```

Query from DBeaver:

```python
SELECT 
	*,
	-- extract age group:
	CASE
		WHEN ind_type IN ('Y0_15', 'Y16_19', 'Y20_24','Y25_34','Y35_44','Y45_54','Y55_64','Y65_74','Y75_MAX')
	END AS gender,
	
	-- extract education:
	CASE
		WHEN ind_type IN ('I0','I3','I5')
	END AS edu_level,
	
	-- extract settlement:
	CASE
		WHEN ind_type IN ('IND_DEG1','IND_DEG2','IND_DEG3')
	END AS settlemnt,
	
	-- extraxt total:
	CASE
		WHEN ind_type = 'IND_TOTAL'
	END AS total,
	
	-- extract education and gender:
	CASE
		WHEN ind_type IN ('M_I0_2', 'M_I3_4', 'M_I5_8')
	END AS male_edu,
	
	CASE
		WHEN ind_type IN ('F_I0_2','F_I3_4', 'F_I5_8')
	END AS fem_edu,
	
	-- extract age + low education:
	CASE
		WHEN ind_type IN ('Y16_24LO','Y16_29LO','Y25_54LO','Y25_64LO')
	END AS age_low_edu
	
	
	
	
	
	
FROM public.stg_dig_skills_baseline
WHERE 
	indicator_code = 'I_DSK2_AB'
	AND "year" = 2025
;




    -- Extract Gender
    CASE 
        WHEN ind_type IN ('M', 'F') THEN ind_type 
        ELSE 'TOTAL' 
    END AS gender,
    
   
FROM clean_data
```

Working version:

SELECT 
	country_code,
	indicator_value,
	ind_type
FROM public.prep_eurostat_filtered
WHERE
	indicator_code = 'I_DSK2_AB'
	AND "year" = 2025
	AND 
	(ind_type IN ('I0','I3','I5') -- getting edu levels
	OR ind_type IN ('IND_DEG1','IND_DEG2','IND_DEG3') -- getting settlement
	OR ind_type = 'IND_TOTAL' -- getting one number per country
	OR ind_type IN ('F_Y16_74', 'M_Y16_74')) -- getting totals for gender
ORDER BY country_code, ind_type;
